Folie 53 : __str__ und __repr__

```python
class Robot:
    def __init__(self, name, __year_of_construction):
        self.__name = name
        self.__year_of_construction = __year_of_construction
   
    def __str__(self):
        return str(self.__dict__)
    
    def __repr__(self):
        return("Robot('"+self.__name+"','"+str(self.__year_of_construction)+"')")
    
    def __eq__(self, other):
        return repr(self)==repr(other)
    
    def evaluation(self):
        return (self.__name + self.__name[::-1])*self.__year_of_construction

r = Robot("Terminator", 2050)
r.evaluation()
```

```python
import random

namen = ("R2D2", "Terminator", "Data")
years_of_construction = (1977, 2029, 2338)
evaluations = {}
amount = 0
for i in range (100):
    n = random.choice(namen)
    b = random.choice(years_of_construction)
    r = Robot(n, b)
    if repr(r) not in evaluations:
        evaluations[repr(r)] = r.evaluation()
    else:        
        amount += 1
print(str(amount) + " Calculations Saved!")
```

Folie 57 Magische Methoden (inkl. Operatorüberladung)

```python
class Length:

    __metric = {"mm" : 0.001, 
                "cm" : 0.01,
                "m"  : 1,
                "km" : 1000,
                "in" : 0.0254,
                "ft" : 0.3048,
                "yd" : 0.9144,
                "mi" : 1609.344 }
    
    def __init__(self, value, unit = "m" ):
        self.value = value
        self.unit = unit
    
    def Converse2Metres(self):
        return self.value * Length.__metric[self.unit]
    
    def __add__(self, other):
        if type(other) == (int or float):
            l = self.Converse2Metres() + other
        else:
            l = self.Converse2Metres() + other.Converse2Metres()
        return Length(l / Length.__metric[self.unit], self.unit )

    #falls die rechte Seite der rechnung unsere Klasse ist a = b + Length
    def __radd__(self, other):
           return self.__add__(other)
   
    def __str__(self):
        return str(self.Converse2Metres())
    
    def __repr__(self):
        return str((self.value, self.unit))
    
l = Length(4,"yd")
r = Length(1000, "cm")
print(l+r)
l = l + 1
print(l)
l = 9 + l
print(l)
l+=5
print(l)
a = 1
a+=l
print(a)
```

Folie 63 Klassenmethode und Vererbung

```python
class Pets():
    name = "Haustiere"
    counter = 0
    
    def __init__(self):
        type(self).counter += 1
        if type(self) != Pets:
            Pets.counter += 1        
            
    def __del__(self):
        type(self).counter -= 1
        if type(self) != Pets:
            Pets.counter -= 1 

    @classmethod
    def about(cls):
        print("{} sind {}".format(cls.counter, cls.name))

class Cats(Pets):
    name = "Katzen"
    counter = 0

    
class Dogs(Pets):
    name = "Hunde"
    counter = 0
```

Folie 68

```python
class Singleton(type):

    _instances = {}
    
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = type.__call__(cls, *args, **kwargs)
        return cls._instances[cls]


class SingletonClass(metaclass=Singleton):
    pass

class NoSingletonClass():
    pass


x = SingletonClass()
y = SingletonClass()
print(x == y, id(x), id(y))

l = NoSingletonClass()
m = NoSingletonClass()
print(l == m, id(l), id(m))
```
